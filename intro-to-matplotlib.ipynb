# Matplotlib Example Workflow

```python
# import matplotlib and get it ready for plotting in jupyter
%matplotlib inline
import matplotlib.pyplot as plt

# 1. prepare data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 10))  ##(10, 10) is a (width, height)

# 3. plot data
ax.plot(x, y)

# 4. Customize plot
ax.set(title="Sample plot",
      xlabel="x-axis",
      ylabel="y-axis")

# 5. Save and show(you save the whole figure)
fig.savefig("C:/Users/RAJA KANNAN/Desktop/udemy machine learning/Images/sample-plot.png")
```

# Making Figures with Numpy arrays

we want:

* line plot
* Scatter plot
* Bar plot
* Histogram
* Sub plots

```python
import numpy as np
```

```python
# Create some data
x = np.linspace(0, 10, 100)   ##linspace allow u to create evenly allowed spaced arrays with specified number of elements
x[:10]
```

```python
# plot the data and create a line plot
fig, ax = plt.subplots()
ax.plot(x, x**2);
```

```python
# use same data to make a scatter
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
```

```python
# Another scatter plot
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x));
```

```python
# Make a plot from dictionary
nut_butter_prices = {"Almond butter":10,
                    "Peanut butter":8,
                    "Cashew butter":12}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax.set(title="Rk's Nut Butter Store",
      ylabel= "Prices");
```

```python
fig, ax = plt.subplots()
ax.barh(list(nut_butter_prices.keys()), nut_butter_prices.values());  ## h=horizontal
```

```python
# Make some data for histogram and plot it
x = np.random.randn(1000)  ##numpy.random.randn() generates random numbers from a standard normal distribution
fig, ax = plt.subplots()
plt.hist(x);
```

# Two options for subplot

```python
# Subplot option 1
fig, ((ax1, ax2),(ax3, ax4)) = plt.subplots(nrows=2,
                                            ncols=2,
                                            figsize=(10, 5))
# Plot to each different axis
ax1.plot(x, x/2);
ax2.scatter(np.random.random(10), np.random.random(10));   ##random.random() - Generates floating-point numbers.
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax4.hist(np.random.randn(1000));
```

```python
# Subplot option 2
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10, 5))
# Plot to each different index
ax[0, 0].plot(x, x/2);
ax[0, 1].scatter(np.random.random(10), np.random.random(10));   ##random.random() - Generates floating-point numbers.
ax[1, 1].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 0].hist(np.random.randn(1000));
```

# Which one you should use?(pyplot vs matplotlib OO method?)
    -When plotting something quickly, okay to use pyplot method
    -When plotting something more advanced, use the OO method

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"]>50]
over_50.head()
```

```python
#pyplot method
over_50.plot(kind="scatter",
            x="age",
            y="chol",
            c="target");  ##c is the color bar
```

```python
## OO method
fix, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind='scatter',
            x='age',
            y='chol',
            c='target',
            ax=ax);
```
